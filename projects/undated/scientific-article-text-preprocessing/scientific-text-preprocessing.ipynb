# Text Preprocessing for Text Mining

Text preprocessing is a crucial step in text mining, as it helps to clean, organize, and prepare textual data for analysis. Below are fundamental text preprocessing steps to prepare raw text data for various machine learning and text mining tasks.

````python
import json
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from sklearn.feature_extraction.text import CountVectorizer

# Download stopwords and lemmatizer data if needed
nltk.download('stopwords')
nltk.download('wordnet')
````

````python

# Initialize pre-processing tools
stop_words = set(stopwords.words("english"))
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()
````

````python

# Load the JSON file
with open("json", "r") as file:
    data = json.load(file)
print(data[3])
````


---

## 1. Tokenization

**Definition**: Tokenization is the process of breaking down text into smaller pieces, typically words or sentences, called tokens. In most cases, tokens are individual words, but they can also represent sentences or phrases.

**Importance**: Tokenization enables easy analysis by converting the text into manageable units. It forms the basis for further text-processing tasks like stop-word removal, stemming, and lemmatization.


````python

# 1. Tokenization
def tokenize(text):
    return nltk.word_tokenize(text)
````

## 2. Stop Words Removal

**Definition**: Stop words are common words in a language that often carry little meaning, such as "is," "and," "the," etc. They are typically removed to focus on more meaningful words in the text.

**Importance**: Removing stop words reduces the dataset’s dimensionality, which can lead to faster processing and improved model performance. It also helps highlight keywords that are more relevant to the content.

**Examples of Common Stop Words**:
- English stop words: "a," "an," "the," "and," "in," "for," "is"
  
**Implementation**:
Most NLP libraries, such as NLTK or spaCy, offer predefined lists of stop words that can be used to filter out these words.


````python

# 2. Stop Words Removal
def remove_stopwords(tokens):
    return [word for word in tokens if word.lower() not in stop_words]
````

## 3. Stemming vs. Lemmatization

**Stemming**: This process reduces a word to its root form by removing suffixes. Stemming algorithms, like the Porter Stemmer, can be aggressive and may not produce actual words.
  
- Example: "running" → "run," "studies" → "studi"

**Lemmatization**: This process converts words to their base or dictionary form (lemma) by considering the context and part of speech. It uses a lexicon to ensure the resulting root form is a valid word.
  
- Example: "running" → "run," "studies" → "study"

**Which to Use**:
For sentiment analysis, **lemmatization** is often preferred because it retains the meaning of words better than stemming, which may be too aggressive and introduce distortions.

---

````python

# 3. Stemming
def stem_words(tokens):
    return [stemmer.stem(word) for word in tokens]
````

````python

# 4. Lemmatization
def lemmatize_words(tokens):
    return [lemmatizer.lemmatize(word) for word in tokens]
````


## 4. Regular Expressions

**Definition**: Regular expressions (regex) are patterns used to match specific sequences of characters in text. They are essential for finding and replacing patterns, like email addresses, dates, or specific symbols, in text preprocessing.

**Use Cases in Text Preprocessing**:
- Removing special characters or HTML tags
- Identifying and removing URLs from text
- Extracting specific patterns, like dates or phone numbers
To remove all non-alphanumeric characters (e.g., punctuation) from a sentence:
```python
import re
text = "CO2-absorbent ceramics improve efficiency!"
cleaned_text = re.sub(r"[^a-zA-Z0-9\s]", "", text)
# Output: "CO2absorbent ceramics improve efficiency"

````python
# 5. Regular Expressions - Removing Non-Alphanumeric Characters
def remove_special_characters(text):
    return re.sub(r'[^a-zA-Z0-9\s]', '', text)
````

## 6. Case Normalization

**Definition**: Case normalization is the process of converting all text to the same case (usually lowercase) to ensure that words like "CO2" and "co2" are treated identically.

**Importance**: This step reduces redundancy in text data, as case variations (e.g., "The" vs. "the") can create unnecessary distinctions. It helps simplify the analysis and often improves model performance.
  
Example: Converting "CO2 Absorbent Ceramics" to lowercase:
- "CO2 Absorbent Ceramics" → "co2 absorbent ceramics"


````python

# 6. Case Normalization
def normalize_case(text):
    return text.lower()
````

````python

# 7. Feature Extraction - Bag of Words
def bag_of_words(texts):
    vectorizer = CountVectorizer()
    X = vectorizer.fit_transform(texts)
    return X.toarray(), vectorizer.get_feature_names_out()
````

````python

# Process each article in the dataset
processed_articles = []
for article in data:
    title = article.get("Title", "")
    abstract = article.get("Abstract", "")
    
    # Combine title and abstract for processing
    combined_text = f"{title} {abstract}"
    
    # Apply preprocessing steps
    combined_text = normalize_case(combined_text)  # Case normalization
    combined_text = remove_special_characters(combined_text)  # Remove special characters
    tokens = tokenize(combined_text)  # Tokenize
    tokens = remove_stopwords(tokens)  # Remove stop words
    stemmed_tokens = stem_words(tokens)  # Stemming
    lemmatized_tokens = lemmatize_words(tokens)  # Lemmatization
    
    # Save processed data
    processed_articles.append({
        "Original Title": title,
        "Original Abstract": abstract,
        "Tokens": tokens,
        "Stemmed Tokens": stemmed_tokens,
        "Lemmatized Tokens": lemmatized_tokens
    })

# Optional: Feature extraction with Bag of Words on entire abstracts collection
abstract_texts = [article["Abstract"] for article in data if article.get("Abstract")]
bow_matrix, feature_names = bag_of_words(abstract_texts)
````

````python

# Display results
print("Processed Articles Sample:")
for i, article in enumerate(processed_articles[:40]):  # Display 40 for brevity
    print(f"\n\nArticle {i+1}\n\n")
    print("Original Title:", article["Original Title"])
    print("\nTokens:", article["Tokens"])
    print("\nStemmed Tokens:", article["Stemmed Tokens"])
    print("\nLemmatized Tokens:", article["Lemmatized Tokens"])

print("\nBag of Words Matrix:\n", bow_matrix[:40])  # Display first 40 rows for brevity
print("Feature Names:", feature_names)
````
